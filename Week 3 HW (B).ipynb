```python
from transformers import pipeline
```

```python
generator = pipeline(
    "text-generation",
    model="google/gemma-2b-it",
    device_map="auto"
)
```

```python
persona = """
You are Gamer Sai.

You are energetic, funny, dramatic, and obsessed with gaming.

You compare:
- school to side quests
- exams to boss fights
- homework to XP grinding
- mathematics to horror survival games

You use gamer slang naturally.
"""
```

```python
def chat(user_input):

    prompt = f"""
{persona}

User: {user_input}
AI:
"""

    output = generator(
        prompt,
        max_new_tokens=80,
        temperature=0.7,
        do_sample=True,
        repetition_penalty=1.1,
        return_full_text=False
    )

    print(output[0]["generated_text"])
```

```python
chat("What do you think about exams?")
```

```python
chat("How do gamers survive school?")

chat("Why is maths terrifying?")

chat("What scares you?")

chat("Should students rage quit homework?")
```

```python
chat("Explain calculus")

chat("Write a formal business email")

chat("Explain quantum physics")
```

```python
gpt2 = pipeline(
    "text-generation",
    model="gpt2"
)
```

```python
prompt = persona + "\nUser: Tell me about school\nAI:"

output = gpt2(
    prompt,
    max_new_tokens=80,
    do_sample=True,
    temperature=0.8
)

print(output[0]['generated_text'])
```
